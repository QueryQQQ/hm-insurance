```python
from dataclasses import dataclass

from pydantic import BaseModel, ConfigDict


# pydantic basemodel类型
class Userresponse(BaseModel):
    model_config = ConfigDict(from_attributes=True)

    id: int
    user: str
    age: int


# # 普通类    
# class User:
#     # init魔法方法负责将输入进来的数据像填表格一样填到规定对象里
#     def __init__(self, id, user, age):
#         self.id=id,
#         self.user=user,
#         self.age=age,

# 普通类  
# dataclass会直接给普通类提供各种魔法函数
@dataclass  
class User:
        id: int
        user: str
        age: int
                                                                                                                        
user = User(id=1, user="jack", age=20)

user_response = Userresponse.model_validate(user)

print(type(user_response))
print(user_response)
```

然而，在很多地方需要我们手动把数据查询出来的普通类对象转为BaseModel类对象，而这都需要通过下面的步骤：

1. 给BaseModel类对象添加属性： model_config = ConfigDict(from_attributes=True) （前提是两种类型属性名是一样的）

2. 用Pydantic BaseModel提供的 model_validate 方法转换对象
