```python

# langchain模型调用四步走
# 1. 读取环境变量
# 环境变量本质就是一串 "KEY=VALUE" 形式的字符串。C 语言里那个变量就叫 char **environ，一个字符串数组。Python 在上面套了个字典样的壳，让你能方便地按名字取
from dotenv import load_dotenv

load_dotenv()# 读取env，并使其加入到环境变量中

# 2. 定义工具,就是定义函数,给函数加装饰器@tool
from langchain.tools import tool


@tool
def get_weather(location: str) -> str:
    """search the current weather in a given location"""
    return f"It's always sunny in {location}!"

@tool
def square_root(number: float) -> float:
    """calculate the square root of a number"""
    return number ** 0.5

# 3. 定义agent，绑定模型和工具
from langchain.agents import create_agent

agent = create_agent(
    model="deepseek-flash",
    tools = [get_weather,square_root]
    )

# 4.调用agent,调用invoke,传入Dict,字典要有message属性,里面是消息的列表
response = agent.invoke({"messages": [{"role":"user", "content": "What's the weather in Guangzhou?"}]})# invoke调用v.
print(response)




```

```python
print(response["messages"][-1].content)# 由于是列表我们只想得到列表最后一条,加-1索引
```
