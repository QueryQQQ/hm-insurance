```python


from dotenv import load_dotenv
from langchain.agents import create_agent
from langchain.tools import tool
from langchain_ollama import ChatOllama

# 1.加载环境变量
load_dotenv()# 读取env，并使其加入到环境变量中

# 2.定义工具
@tool
def get_weather(location: str) -> str:
    """search the current weather in a given location"""
    return f"It's always sunny in {location}!"

# 3.定义模型,自定义初始化,就是手动传模型的名字,APIKey和BaseURL
model = ChatOllama(
    model = "qwen3.5:0.8b",
    temperature = 0.8,
)

# 4.定义agent,绑定模型和工具    
agent = create_agent(
    model=model,
    tools=[get_weather],
)

# 5.调用agent
response = agent.invoke({"messages": [{"role": "user", "content": "你是谁?"}]})
print(response)
```
