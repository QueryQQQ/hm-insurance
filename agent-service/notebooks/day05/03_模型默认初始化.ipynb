```python
# 1.加载环境变量
from dotenv import load_dotenv

load_dotenv()# 读取env，并使其加入到环境变量中

# 2.定义工具
from langchain.tools import tool


@tool
def get_weather(location: str) -> str:
    """search the current weather in a given location"""
    return f"It's always sunny in {location}!"

# 3.定义模型默认初始化,就是只传名字即可(只支持deepseek,其余传api),也可以加上模型自身参数,系统自动在env里推断出该用的api
from langchain.chat_models import init_chat_model

model = init_chat_model(
    "deepseek-flash",
    temperature = 0.2,
    extra_body = {"thinking": {"type": "disabled"}},# thinking type enabled/disabled
)

# 4.定义agent,绑定模型和工具    
from langchain.agents import create_agent

agent = create_agent(
    model=model,
    tools=[get_weather],
)

# 5.调用agent
response = agent.invoke({"messages": [{"role": "user", "content": "广州天气怎么样?"}]})
print(response)
```
